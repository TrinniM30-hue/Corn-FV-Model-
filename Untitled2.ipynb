```python
import numpy as np
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from matplotlib.cm import get_cmap
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
import datetime as dt
```

```python
df = pd.read_csv("usda_data.csv")

df["Commodity"] = df["Commodity"].ffill()

year_cols = [col for col in df.columns if '/' in col]

for col in year_cols:
    df[col] = pd.to_numeric(
        df[col].astype(str).str.replace(',', '', regex=False)
    )
    
df_long = df.melt(
    id_vars = ['Commodity', 'Attribute', 'Country', 'Unit Description'],
    value_vars = year_cols, 
    var_name = 'Marketing_Year',
    value_name = 'Value'

)

def extract_start_year(marketing_year_str):
    return int(marketing_year_str.split('/')[0])

df_long['Year'] = df_long['Marketing_Year'].apply(extract_start_year)

df_long.isnull().sum()

df_long.head(10) 
```

```python
start_date = "2000-01-01"

corn = yf.download("ZC=F", start=start_date, auto_adjust=True)
soy = yf.download("ZS=F", start=start_date, auto_adjust=True)

corn_close = corn[('Close', 'ZC=F')].rename('Corn_Close')
soy_close = soy[('Close', 'ZS=F')].rename('Soy_Close')

corn_monthly = corn_close.resample('ME').last().reset_index()
soy_monthly  = soy_close.resample('ME').last().reset_index()

corn_monthly['Year'] = corn_monthly['Date'].dt.year
soy_monthly['Year']  = soy_monthly['Date'].dt.year

futures = corn_monthly.merge(soy_monthly[['Date', 'Soy_Close']], on='Date', how='inner')

usda_pivot = df_long.pivot_table(
    index=['Commodity', 'Year'],
    columns='Attribute',
    values='Value',
    aggfunc='first'
).reset_index()
usda_pivot.columns.name = None


usda_pivot['Stocks_to_Use'] = (
    usda_pivot['Ending Stocks'] /
    (usda_pivot['Exports'] + usda_pivot['Domestic Consumption']))
    
corn_usda = usda_pivot[usda_pivot['Commodity'] == 'Corn'][
    ['Year', 'Stocks_to_Use']
].rename(columns={'Stocks_to_Use': 'Corn_StU'})

soy_usda = usda_pivot[usda_pivot['Commodity'] == 'Oil, Soybean'][
    ['Year', 'Stocks_to_Use']
].rename(columns={'Stocks_to_Use': 'Soy_StU'})
    

usda_final = corn_usda.merge(soy_usda, on='Year', how='inner')

    
usda_final['StU_Ratio'] = usda_final['Soy_StU'] / usda_final['Corn_StU']

#Assigning marketing year for Futures 

def assign_marketing_year(date):
    if date.month >= 9:
        return date.year
    else:
        return date.year - 1

futures['Marketing_Year'] = futures['Date'].apply(assign_marketing_year)

model_df = futures.merge(usda_final, left_on='Marketing_Year', right_on='Year', how='left')

model_df.head(20)
                                
#Marketing Year Avg Spread
```

```python
corn_my = model_df.groupby('Marketing_Year')['Corn_Close'].mean().reset_index()
corn_my.columns = ['Year', 'Corn_Avg']
corn_my['Corn_Avg'] = corn_my['Corn_Avg'] / 100

corn_model = corn_my.merge(usda_final, on='Year', how='inner')
corn_model['ln_Corn'] = np.log(corn_model['Corn_Avg'])
print(corn_model.tail())

X = corn_model[['Corn_StU']]
y = corn_model['ln_Corn']

lm = LinearRegression()
lm.fit(X, y)

alpha = lm.intercept_
beta_corn = lm.coef_[0]
r2 = lm.score(X, y)

print(f"R²:        {r2:.4f}")
print(f"Intercept: {alpha:.4f}")
print(f"Corn_Beta:  {beta_corn:.4f}")

# Add predicted ln_Spread to spread_model
corn_model['FV_ln_Corn'] = lm.predict(X)

# Convert back from log to dollars
corn_model['FV_Corn'] = np.exp(corn_model['FV_ln_Corn'])

# Compute % deviation
corn_model['Pct_Dev'] = (
    (corn_model['Corn_Avg'] - corn_model['FV_Corn']) 
    / corn_model['FV_Corn'] * 100
)

print(corn_model[['Year', 'Corn_Avg', 'FV_Corn', 'Pct_Dev']].tail(6))

current_year = 2025
row = corn_model[corn_model['Year'] == current_year]


print(f"Actual Corn Price:  ${row['Corn_Avg'].values[0]:.2f}/bu")
print(f"Fair Value:     ${row['FV_Corn'].values[0]:.2f}/bu")
print(f"% Deviation:    {row['Pct_Dev'].values[0]:+.1f}%")
```

```python
fig, axes = plt.subplots(2, 1, figsize=(12, 8))

# Panel 1: your existing chart — move it to axes[0]
axes[0].plot(corn_model['Year'], corn_model['Corn_Avg'], label='Actual')
axes[0].plot(corn_model['Year'], corn_model['FV_Corn'], linestyle='--', color='red', label='Fair Value')
axes[0].legend()
axes[0].set_title('Corn Price vs Fair Value')

colors = ['red' if x > 0 else 'green' for x in corn_model['Pct_Dev']]

axes[1].bar(corn_model['Year'], corn_model['Pct_Dev'], color=colors)
axes[1].set_title('Model Deviations\n(Red = Overvalued, Green = Undervalued)')
for year, pct in zip(corn_model['Year'], corn_model['Pct_Dev']):
    if abs(pct) > 0:
        axes[1].text(year, pct + (1 if pct > 0 else -3), 
                     f"{pct:+.0f}%", 
                     ha='center', fontsize=8, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

ax.scatter(corn_model['Corn_StU'], corn_model['Corn_Avg'], 
           color='steelblue', s=80)

for _, row in corn_model.iterrows():
    ax.annotate(str(int(row['Year']))[2:],
                (row['Corn_StU'], row['Corn_Avg']),
                textcoords='offset points', xytext=(5, 5), fontsize=8)
    
    
stu_range = np.linspace(corn_model['Corn_StU'].min(), corn_model['Corn_StU'].max(), 200)
curve = np.exp(alpha + beta_corn * stu_range)
ax.plot(stu_range, curve, color='red', linewidth=2)

ax.set_xlabel('Corn Stocks-to-Use Ratio')
ax.set_ylabel('Corn ($/bu)')
ax.set_title('Corn Price vs Corn S/U')

plt.show()
```
